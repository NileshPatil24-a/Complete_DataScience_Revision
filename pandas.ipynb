```python
df = pd.read_csv("https://raw.githubusercontent.com/AkarshVyas/Pandas-Youtube/refs/heads/main/anime.csv")
```

```python
df.head()
```

```python
def episode(string ):
  return string.split("(")[1][0:3].strip()
```

```python
df["episode"] = df["Title"].apply(episode)
```

```python
def replace_e(txt):
  return int(txt.replace("e", ""))
```

```python
df["episode"] = df["episode"].apply(replace_e)
```

```python
df.head()
```

```python
df["time_stamp"] =  df["Title"].apply(lambda txt : txt.split(")")[1][:19])
```
