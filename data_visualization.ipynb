```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

matplotlib dasics

```python
X = np.linspace(0,5, 11)
```

```python
Y = X**2
```

```python
plt.plot(X,Y)
plt.title("Title")
plt.xlabel("chacha")
plt.ylabel("chachi")
```

Subplots in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(X,Y)
plt.subplot(2,2,4)
plt.plot(X**2,Y)
plt.subplot(2,2,3)
plt.plot(Y,X)
plt.subplot(2,2,2)
plt.plot(Y**2,X)
```

```python
plt.figure(figsize=(5,1))
plt.plot(X,Y)
```

```python
plt.scatter(X,Y)
```

```python
plt.hist(X)
```

```python
plt.boxplot(X)
plt.savefig("sham.png")
```

```python
df = sns.load_dataset("tips")
```

```python
df.head()
```

```python
sns.displot(df["total_bill"],bins=20,kde=True)
```

```python
sns.jointplot(x = "total_bill", y="tip", data=df,kind="scatter")
```

```python
sns.pairplot(df, hue="sex")
```

Categorical plots

```python
sns.countplot(x = df['sex'])
```

```python
sns.barplot(x=df['sex'], y=df['tip'])
```

```python
sns.boxplot(x = df['tip'], y=df['sex'], palette="rainbow")
```
